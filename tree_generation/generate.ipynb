```python
import numpy as np
import matplotlib.pyplot as plt
import networkx as nx

# Parameters
N1 = 900  # Number of nodes in group 1
N2 = 100  # Number of nodes in group 2
N = N1 + N2  # Total number of nodes


# Create random geometric graph
def create_random_geometric_graph(radius):
    x_positions = np.concatenate([np.random.normal(0, 1, N1), np.random.normal(3, 1, N2)])
    y_positions = np.random.normal(0, 0.25, N)

    positions = np.column_stack((x_positions, y_positions))
    G = nx.Graph()
    for i in range(len(positions)):
        G.add_node(i)
        for j in range(i + 1, len(positions)):
            distance = np.linalg.norm(positions[i] - positions[j])
            if distance < radius:
                G.add_edge(i, j)
    return G, positions

import networkx as nx

def save_edge_list_to_txt(G, filename):
    with open(filename, 'w') as file:
        for edge in G.edges():
            file.write(f"{edge[0]} {edge[1]}\n")

# Function to save node positions to a .txt file
def save_positions_to_txt(positions, filename):
    with open(filename, 'w') as file:
        for i, pos in enumerate(positions):
            file.write(f"{i} {pos[0]} {pos[1]}\n")


radii = np.linspace(2, 3, 20)  # Example radius values
graphs = []
positions = []
count = 0
for radius in radii:
    G, position = create_random_geometric_graph(radius)
    graphs.append(G)
    positions.append(position)
    if nx.is_connected(G):
        save_edge_list_to_txt(G, f"test_{count}.txt")
        save_positions_to_txt(position, f"test_cood_{count}.txt")
        print(f"Graph {count} is connected and saved as 'test_{count}.txt'.")
        count += 1
```

```python
# Gradually increase the radius until the graph becomes connected
radius = 0.01  # Initial radius
max_radius = 1.0  # Maximum allowable radius to avoid infinite loop
step = 0.01  # Increment in radius
connected_radius = None

while radius <= max_radius:
    G, position = create_random_geometric_graph(radius)
    if nx.is_connected(G):  # Check if the graph is connected
        connected_radius = radius
        break
    radius += step

if connected_radius is not None:
    print(f"The graph becomes connected at radius: {connected_radius}")
else:
    print("The graph did not become connected within the maximum radius.")

# Plot the graph at the first connected radius
if connected_radius is not None:
    plt.figure(figsize=(8, 8))
    nx.draw(
        G,
        pos={i: pos for i, pos in enumerate(position)},
        node_size=20,
        with_labels=False,
        node_color="blue"
    )
    plt.title(f"Random Geometric Graph (Connected at Radius: {connected_radius})")
    plt.show()
```
